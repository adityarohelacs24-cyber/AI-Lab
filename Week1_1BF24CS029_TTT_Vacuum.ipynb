```python
import random

def display_board(board):
    """Prints the Tic-Tac-Toe board."""
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("-----------")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("-----------")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print("\n")

def player_move(board, player_marker):
    """Handles the player's move."""
    while True:
        try:
            move = int(input(f"Enter your move (1-9) for {player_marker}: ")) - 1
            if 0 <= move <= 8 and board[move] == ' ':
                board[move] = player_marker
                break
            else:
                print("This spot is taken or invalid. Please choose an empty spot (1-9).")
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 9.")

def computer_move(board, computer_marker, player_marker):
    """Handles the computer's move with a simple AI."""

    for i in range(9):
        if board[i] == ' ':
            board[i] = computer_marker
            if check_win(board, computer_marker):
                return
            board[i] = ' '


    for i in range(9):
        if board[i] == ' ':
            board[i] = player_marker
            if check_win(board, player_marker):
                board[i] = computer_marker
                return
            board[i] = ' '


    if board[4] == ' ':
        board[4] = computer_marker
        return


    corners = [0, 2, 6, 8]
    available_corners = [c for c in corners if board[c] == ' ']
    if available_corners:
        board[random.choice(available_corners)] = computer_marker
        return


    sides = [1, 3, 5, 7]
    available_sides = [s for s in sides if board[s] == ' ']
    if available_sides:
        board[random.choice(available_sides)] = computer_marker
        return

def check_win(board, marker):
    """Checks if the given marker has won the game."""
    winning_combinations = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]
    for combo in winning_combinations:
        if board[combo[0]] == board[combo[1]] == board[combo[2]] == marker:
            return True
    return False

def check_draw(board):
    """Checks if the game is a draw."""
    return ' ' not in board

def play_game():
    """Main function to play Tic-Tac-Toe."""
    print("Welcome to Tic-Tac-Toe!")
    board = [' '] * 9

    player_marker = ''
    while not (player_marker == 'X' or player_marker == 'O'):
        player_marker = input("Do you want to be X or O? ").upper()

    computer_marker = 'O' if player_marker == 'X' else 'X'
    print(f"You are {player_marker}. The computer is {computer_marker}.")

    turn = random.choice([player_marker, computer_marker])
    print(f"The first turn goes to {turn}.")

    game_over = False
    while not game_over:
        if turn == player_marker:
            display_board(board)
            player_move(board, player_marker)
            if check_win(board, player_marker):
                display_board(board)
                print("Congratulations! You won!")
                game_over = True
            elif check_draw(board):
                display_board(board)
                print("It's a draw!")
                game_over = True
            else:
                turn = computer_marker
        else:
            print(f"\nComputer ({computer_marker}) is making a move...")
            computer_move(board, computer_marker, player_marker)
            if check_win(board, computer_marker):
                display_board(board)
                print("Computer wins!")
                game_over = True
            elif check_draw(board):
                display_board(board)
                print("It's a draw!")
                game_over = True
            else:
                turn = player_marker

    print("\nGame Over.")


if __name__ == '__main__':
    play_game()
```

```python
import random
import time

class Room:
    def __init__(self, room_id, is_dirty=False):
        self.room_id = room_id
        self.is_dirty = is_dirty

    def __str__(self):
        return f"Room {self.room_id}: {'Dirty' if self.is_dirty else 'Clean'}"

class VacuumCleanerAgent:
    def __init__(self, num_rooms):
        self.location = 0  # Start in Room 0
        self.rooms = [Room(i, is_dirty=(random.random() < 0.5)) for i in range(num_rooms)]
        self.action_cost = 0

    def perceive(self):
        """Perceives the current state of the room."""
        return self.rooms[self.location].is_dirty, self.location

    def reflect_and_act(self, is_dirty, current_location):
        """Simple reflex agent rule: if dirty, clean; otherwise, move."""
        if is_dirty:
            print(f"Agent at Room {current_location} is Dirty. Cleaning...")
            self.rooms[current_location].is_dirty = False
            self.action_cost += 1
            return "CLEAN"
        else:
            # Move to the next room (cyclically)
            next_location = (current_location + 1) % len(self.rooms)
            print(f"Agent at Room {current_location} is Clean. Moving to Room {next_location}.")
            self.location = next_location
            self.action_cost += 1
            return "MOVE"

    def is_goal_state(self):
        """Checks if all rooms are clean."""
        return all(not room.is_dirty for room in self.rooms)

    def run(self, max_steps=100):
        """Runs the vacuum cleaner agent until all rooms are clean or max_steps reached."""
        print("\n--- Initial State ---")
        for room in self.rooms:
            print(room)
        print("---------------------")
        print(f"Agent starts at Room {self.location}\n")

        step = 0
        while not self.is_goal_state() and step < max_steps:
            print(f"--- Step {step + 1} ---")
            is_dirty, current_location = self.perceive()
            action = self.reflect_and_act(is_dirty, current_location)
            for room in self.rooms:
                print(room)
            print(f"Action taken: {action}. Total cost: {self.action_cost}\n")
            time.sleep(0.5) # Simulate real-time actions
            step += 1

        print("--- Final State ---")
        for room in self.rooms:
            print(room)
        if self.is_goal_state():
            print("Goal Achieved! All rooms are clean.")
        else:
            print("Max steps reached. Not all rooms are clean.")
        print(f"Total actions (cost): {self.action_cost}")

# --- Program Execution ---
if __name__ == '__main__':
    num_rooms_to_clean = int(input("Enter the number of rooms (n): "))
    if num_rooms_to_clean <= 0:
        print("Number of rooms must be positive.")
    else:
        agent = VacuumCleanerAgent(num_rooms_to_clean)
        agent.run()
```
