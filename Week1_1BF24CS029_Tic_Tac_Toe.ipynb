```markdown
# Tic-Tac-Toe Game
```

````python
import random

def display_board(board):
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("-----------")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("-----------")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print("\n")

def check_win(board, player):
    for i in range(0, 9, 3):
        if board[i] == board[i+1] == board[i+2] == player:
            return True
    for i in range(3):
        if board[i] == board[i+3] == board[i+6] == player:
            return True
    if board[0] == board[4] == board[8] == player:
        return True
    if board[2] == board[4] == board[6] == player:
        return True
    return False

def check_draw(board):
    for i in range(9):
        if board[i] == ' ':
            return False
    return True

def get_player_choice(board, player_symbol):
    while True:
        try:
            position = int(input(f"Player {player_symbol}, enter your move (1-9): "))
            if 1 <= position <= 9:
                if board[position-1] == ' ':
                    return position - 1
                else:
                    print("That position is already occupied. Please choose another.")
            else:
                print("Invalid input. Please enter a number between 1 and 9.")
        except ValueError:
            print("Invalid input. Please enter a number.")

board = [' '] * 9
player_symbols = {'human': 'X', 'computer': 'O'}
current_player = 'human'

print("Welcome to Tic-Tac-Toe! You are 'X', the computer is 'O'.")

game_over = False
while not game_over:
    display_board(board)

    player_symbol = player_symbols[current_player]

    if current_player == 'human':
        move_index = get_player_choice(board, player_symbol)
    else:
        print(f"Player {player_symbol} (Computer) is making a move...")
        available_moves = [i for i, spot in enumerate(board) if spot == ' ']
        if available_moves:
            move_index = random.choice(available_moves)
        else:
            game_over = True
            continue

    board[move_index] = player_symbol

    if check_win(board, player_symbol):
        display_board(board)
        print(f"Congratulations, Player {player_symbol}! You win!")
        game_over = True
    elif check_draw(board):
        display_board(board)
        print("It's a draw!")
        game_over = True
    else:
        current_player = 'computer' if current_player == 'human' else 'human'

print("Game Over.")
````
