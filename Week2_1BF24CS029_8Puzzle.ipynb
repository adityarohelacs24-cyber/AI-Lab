## 8-Puzzle Solver using Iterative Deepening Search (IDS)

```python
import collections

class PuzzleState:
    def __init__(self, board, parent=None, move=None, depth=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth
        self.empty_pos = self._find_empty()
    def _find_empty(self):
        for r in range(3):
            for c in range(3):
                if self.board[r][c] == 0:
                    return r, c
        return None
    def __eq__(self, other):
        return self.board == other.board
    def __hash__(self):
        return hash(self.board)
    def __str__(self):
        return '\n'.join([' '.join(map(str, row)) for row in self.board])
    def get_neighbors(self):
        neighbors = []
        er, ec = self.empty_pos
        moves = [(0, 1, 'Right'), (0, -1, 'Left'), (1, 0, 'Down'), (-1, 0, 'Up')]
        for dr, dc, move_name in moves:
            nr, nc = er + dr, ec + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_board_list = [list(row) for row in self.board]
                new_board_list[er][ec], new_board_list[nr][nc] = new_board_list[nr][nc], new_board_list[er][ec]
                new_board_tuple = tuple(tuple(row) for row in new_board_list)
                neighbors.append(PuzzleState(new_board_tuple, self, move_name, self.depth + 1))
        return neighbors
def print_solution(state):
    path = []
    while state:
        path.append(state)
        state = state.parent
    for s in reversed(path):
        if s.move:
            print(f"Move: {s.move}")
        print(s)
        print()
```

### Depth-Limited Search (DLS) Algorithm

This function performs a Depth-Limited Search, which is a component of Iterative Deepening Search. It explores states up to a certain depth limit.

```python
def dls(initial_state, goal_state, limit):
    stack = collections.deque()
    stack.append(initial_state)
    visited = {initial_state}
    while stack:
        current_state = stack.pop()
        if current_state == goal_state:
            return current_state
        if current_state.depth < limit:
            for neighbor in reversed(current_state.get_neighbors()):
                    visited.add(neighbor)
                    stack.append(neighbor)
    return None
```

### Iterative Deepening Search (IDS) Algorithm

This function combines DLS with increasing depth limits to find the shortest path to the goal state.

```python
def ids(initial_board, goal_board):
    initial_state = PuzzleState(initial_board)
    goal_state = PuzzleState(goal_board)
    if initial_state == goal_state:
        return initial_state
    for limit in range(50):
        result = dls(initial_state, goal_state, limit)
        if result is not None:
            return result
    return None
```

### Example Usage

Define an initial and goal state for the 8-puzzle and see the solution path.

```python
initial_board = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)
goal_board = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)
print("Initial State:")
print(PuzzleState(initial_board))
print("\nGoal State:")
print(PuzzleState(goal_board))
print("\nSolving...")
solution = ids(initial_board, goal_board)
if solution:
    print("\nSolution Found!")
    print_solution(solution)
else:
    print("\nNo solution found within the given depth limit.")
```
