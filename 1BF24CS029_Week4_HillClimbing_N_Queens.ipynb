```python
import random
import ast
def calculate_cost(state):
    n = len(state)
    collisions = 0
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                collisions += 1
            if abs(state[i] - state[j]) == abs(i - j):
                collisions += 1
    return collisions
def print_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print()
def get_neighbors_with_info(state):
    neighbors = []
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            neighbor = list(state)
            neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
            neighbors.append((neighbor, (i, j)))
    return neighbors
def hill_climbing_n_queens(initial_state):
    current_state = list(initial_state)
    step = 0
    print(f"--- Initial State ---")
    print_board(current_state)
    current_cost = calculate_cost(current_state)
    print(f"Initial Cost: {current_cost}\n")
    while True:
        if current_cost == 0:
            print("Goal state reached successfully!")
            break
        neighbors = get_neighbors_with_info(current_state)
        if not neighbors:
            break
        best_neighbor = None
        best_cost = current_cost
        best_swap = None
        print("Evaluating all neighbor swaps, their board states, and their costs:\n")
        for neighbor, swap in neighbors:
            cost = calculate_cost(neighbor)
            print(f"Swap columns {swap[0]} and {swap[1]} -> State: {neighbor} | Cost: {cost}")
            print_board(neighbor)
            if cost < best_cost:
                best_cost = cost
                best_neighbor = neighbor
                best_swap = swap
        if best_cost >= current_cost:
            print(f"Stuck in local optimum. No swap improves the current cost of {current_cost}.")
            break
        current_state = best_neighbor
        current_cost = best_cost
        step += 1
        print(f"--- Step {step} (Chosen Best Swap: columns {best_swap[0]} and {best_swap[1]}) ---")
        print_board(current_state)
        print(f"Current Cost: {current_cost}\n")
try:
    user_input = input("Enter the initial state as an array (e.g., [1, 3, 0, 2]): ")
    initial_state = ast.literal_eval(user_input)
    if isinstance(initial_state, list):
        hill_climbing_n_queens(initial_state)
    else:
        print("Please enter a valid list structure, e.g., [1, 3, 0, 2]")
except Exception as e:
    print(f"Error parsing input: {e}. Please use format like [1, 3, 0, 2]")
```
